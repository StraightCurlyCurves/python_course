## Read a file

With the `open` function, you can open a file. The `open` function returns a file object that you can use to read the file.

```python
file = open('message.txt')
print(file.readline().strip()) # read the first line, strip() removes the newline character at the end of the line
print(file.readline().strip()) # read the next line, strip() removes the newline character at the end of the line
print(file.readline().strip()) # read the next line, strip() removes the newline character at the end of the line

# or read all the lines at once and store them in a list
file.seek(0) # move the cursor to the beginning of the file
lines = file.readlines()
print(lines)
file.close() # close the file so that it can be used by other programs
```

## `With` operator

The `with` operator is used to open a file, do operations on it, and automatically close it when the block is done.

It is good practice to use the `with` operator when working with files to avoid leaving files open.

```python
# With operator
with open('message.txt') as file:
    full_content = file.read()
    print(full_content)
```

## Write a file

Analog to the `read` method, there is a `write` method. You have to open the file in write mode!

```python
first_line = 'Hello, World!'
second_line = 'How are you today?'

with open('message.txt', 'w') as file:
    file.write(first_line)
    file.write(second_line)
```

Check out the file. What do you see?

Fix it! A new line is made with the new line [escape code](https://en.wikipedia.org/wiki/ANSI_escape_code) `"\n"`.

## Extend file

To extend a file, open an existing file in write mode and

```python
# write some text at the end of the existing txt file
lines = ['I am doing great!\n', 'How about you?\n']

with open('my_message.txt', 'a') as file:
    file.writelines(lines)
```

## File Methods

| Method                | Description                                                                 |
|-----------------------|-----------------------------------------------------------------------------|
| `file.read(size)`     | Reads up to `size` bytes from the file. If `size` is omitted, reads the entire file. |
| `file.readline()`     | Reads a single line from the file.                                          |
| `file.readlines()`    | Reads all lines from the file and returns a list of strings.                |
| `file.write(string)`  | Writes the string to the file.                                              |
| `file.writelines(lines)` | Writes a list of strings to the file. Each string is written as a separate line. |
| `file.close()`        | Closes the file.                                                            |
| `file.flush()`        | Flushes the internal buffer, writing all buffered data to the file.         |
| `file.seek(offset, whence)` | Moves the file pointer to a specified location. `whence` can be `0` (start), `1` (current), or `2` (end). |
| `file.tell()`         | Returns the current file pointer position.                                  |
| `file.truncate(size)` | Truncates the file to a specified size. If `size` is omitted, truncates to the current file pointer position. |
| `file.mode`           | Returns the mode in which the file was opened.                              |
| `file.name`           | Returns the name of the file.                                               |
| `file.closed`         | Returns `True` if the file is closed, `False` otherwise.                    |
