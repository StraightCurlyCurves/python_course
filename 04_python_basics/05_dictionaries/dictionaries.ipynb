# Introduction to Dictionaries
Dictionaries are used to store data values in `key:value` pairs. A dictionary is a collection which is ordered, changeable (mutable) and does not allow duplicates.

## Overview / Creating a Dictionary

```python

person = {
    "name": "John",
    "age": 30,
    "city": "New York"
}

person2 = {}
person2["name"] = "Jane"
person2["age"] = 25
person2["city"] = "San Francisco"

# Accessing items
print(person["name"])

# Changing values
person["age"] = 31

# Loop through a dictionary
for key in person:
    print(key, person[key])

# Check if a key exists
if "city" in person:
    print("City is present in the dictionary")

# Dictionary length
print(len(person))

# Adding items
person["email"] = "john@example.com"

# Removing items
person.pop("age")

# Clearing the dictionary
person.clear()

# Deleting the dictionary
del person
```

## Accessing Dictionary Items
Access items in a dictionary using the keys.

```python
# Accessing Dictionary Items

# Access items in a dictionary using the keys
car = {
    "brand": "Ford",
    "model": "Mustang",
    "year": 1964
}

# Accessing the value with the [] operator
print(car["brand"])  # Output: Ford
print(car["model"])  # Output: Mustang
print(car["year"])  # Output: 1964

# Using the get() method to access items
print(car.get("brand"))  # Output: Ford
print(car.get("model"))  # Output: Mustang
print(car.get("year"))   # Output: 1964
```

### Difference of behavior between get() and [] operator
If the key does not exist, the [] operator will raise an error while the get() method will return None.

```python
print(car.get("color"))  # Output: None
print(car["color"])  # KeyError: 'color'
```

## Modifying Dictionary Items
Modify items in a dictionary by accessing them using their keys.

```python
car = {
    "brand": "Ford",
    "model": "Mustang",
    "year": 1964
}

# Modifying the value of the key "year"
car["year"] = 2020

# Using the update() method to modify multiple items
car.update({"year": 2021, "color": "blue"})

# Printing the modified dictionary
print(car)  # Output: {'brand': 'Ford', 'model': 'Mustang', 'year': 2021, 'color': 'blue'}
```

## Looping Through a Dictionary
Loop through a dictionary using a for loop to access keys and values.

```python
person = {
    "name": "Alice",
    "age": 28,
    "city": "Los Angeles"
}

# Loop through keys
for key in person:
    print(key)  # Output: name, age, city

# Loop through values
for value in person.values():
    print(value)  # Output: Alice, 28, Los Angeles

# Loop through key-value pairs
for key, value in person.items():
    print(key, value)  # Output: name Alice, age 28, city Los Angeles
```

## Dictionary Methods
Common dictionary methods include:

| Method     | Description                                                                     |
|------------|---------------------------------------------------------------------------------|
| clear()    | Removes all the elements from the dictionary                                    |
| copy()     | Returns a copy of the dictionary                                                |
| fromkeys() | Returns a dictionary with the specified keys and value                          |
| get()      | Returns the value of the specified key                                          |
| items()    | Returns a list containing a tuple for each key value pair                       |
| keys()     | Returns a list containing the dictionary's keys                                 |
| pop()      | Removes the element with the specified key                                      |
| popitem()  | Removes the last inserted key-value pair                                        |
| setdefault()| Returns the value of the specified key. If the key does not exist: insert the key, with the specified value |
| update()   | Updates the dictionary with the specified key-value pairs                       |
| values()   | Returns a list of all the values in the dictionary                              |

```python
person = {
    "name": "Alice",
    "age": 28,
    "city": "Los Angeles"
}

# clear() method - Removes all elements from the dictionary
person.clear()
print(person)  # Output: {}

# Re-creating the dictionary for further examples
person = {
    "name": "Alice",
    "age": 28,
    "city": "Los Angeles"
}

# copy() method - Returns a copy of the dictionary
person_copy = person.copy()
print(person_copy)  # Output: {'name': 'Alice', 'age': 28, 'city': 'Los Angeles'}

# fromkeys() method - Creates a dictionary from the given keys and value
keys = ('name', 'age', 'city')
value = None
new_dict = dict.fromkeys(keys, value)
print(new_dict)  # Output: {'name': None, 'age': None, 'city': None}

# get() method - Returns the value of the specified key
name = person.get("name")
print(name)  # Output: Alice

# items() method - Returns a list containing a tuple for each key-value pair
items = person.items()
print(items)  # Output: dict_items([('name', 'Alice'), ('age', 28), ('city', 'Los Angeles')])

# keys() method - Returns a list containing the dictionary's keys
keys = person.keys()
print(keys)  # Output: dict_keys(['name', 'age', 'city'])

# pop() method - Removes the element with the specified key
age = person.pop("age")
print(age)  # Output: 28
print(person)  # Output: {'name': 'Alice', 'city': 'Los Angeles'}

# popitem() method - Removes the last inserted key-value pair
last_item = person.popitem()
print(last_item)  # Output: ('city', 'Los Angeles')
print(person)  # Output: {'name': 'Alice'}

# Re-creating the dictionary for further examples
person = {
    "name": "Alice",
    "age": 28,
    "city": "Los Angeles"
}

# setdefault() method - Returns the value of the specified key. If the key does not exist, insert the key with the specified value
age = person.setdefault("age", 30)
print(age)  # Output: 28
country = person.setdefault("country", "USA")
print(country)  # Output: USA
print(person)  # Output: {'name': 'Alice', 'age': 28, 'city': 'Los Angeles', 'country': 'USA'}

# update() method - Updates the dictionary with the specified key-value pairs
person.update({"age": 29, "email": "alice@example.com"})
print(person)  # Output: {'name': 'Alice', 'age': 29, 'city': 'Los Angeles', 'country': 'USA', 'email': 'alice@example.com'}

# values() method - Returns a list of all the values in the dictionary
values = person.values()
print(values)  # Output: dict_values(['Alice', 29, 'Los Angeles', 'USA', 'alice@example.com'])
```
